```python
# agg method
# passing dict

genres.agg(
    {
        'Runtime':'mean',
        'IMDB_Rating':'mean',
        'No_of_Votes':'sum',
        'Gross': 'sum',
        'Metascore':'min'
    }
)
```

```python
# passing list

genres.agg(['min','max','mean','sum'])
```

```python
# Adding both the syntax

genres.agg(
    {
        'Runtime':['min','mean'],
        'IMDB_Rating':'mean',
        'No_of_Votes':['sum','max'],
        'Gross': 'sum',
        'Metascore':'min'
    }
)
```

```python
# Looping on groups

# Find the highest rated movie of each genre

result = []

for group, data in genres:
    result.append(data[data['IMDB_Rating'] == data['IMDB_Rating'].max()])

df = pd.concat(result)

df
```

```python
# split (apply) combine
# apply  -> button function

# find no of movies starting with A for each group

def foo(group):
    return group['Series_Title'].str.startswith('A').sum()
```

```python
genres.apply(foo)
```

```python
# find ranking of each movie in the group according to IMDB_Rating

def rank_movies(group):
    group['genre_rank']=group['IMDB_Rating'].rank(ascending=False)
    return group
```

```python
genres.apply(rank_movies)
```

```python
# find normalized IMDB rating group wise

def norm_movies(group):
    group['norm_rating'] = (group['IMDB_Rating'] - group['IMDB_Rating'].min())/(group['IMDB_Rating'].max() - group['IMDB_Rating'].min())
    return group
```

```python
genres.apply(norm_movies)
```

```python
ipl = pd.read_csv('D:\mca_aiml_lab\mca_env\Datasets\deliveries.csv')
```

```python
ipl.head()
```

```python
# Find the top 10 batsman in term of runs

batsman = ipl.groupby('batsman') 

batsman['batsman_runs'].sum().sort_values(ascending=False).head(10)
```

```python
# find the batsman with max no. of sixes

six = ipl[ipl['batsman_runs']==6]

six.groupby('batsman')['batsman_runs'].count().sort_values(ascending=False).head()
```

```python
fourAndSix =ipl[(ipl['batsman_runs'] == 4 ) | (ipl['batsman_runs']==6)]

fourAndSix.head()
```

```python
last4_over = fourAndSix[fourAndSix['over'] > 15]

last4_over.groupby('batsman')['batsman'].count().sort_values(ascending=False).head()
```

```python
# Find V kholi's record against all team

temp_df = ipl[ipl['batsman'] == 'V Kohli']


temp_df.groupby('bowling_team')['batsman_runs'].sum().reset_index()
```

```python
# Create a function that can return the highest score of any batsman

def highest(batsman):
    temp_df = ipl[ipl['batsman'] == batsman]
    return temp_df.groupby('match_id')['batsman_runs'].sum().sort_values(ascending=False).head(1).values[0]

```

```python
print(highest('RG Sharma'))
```

```python
print(highest('CH Gayle'))
```
