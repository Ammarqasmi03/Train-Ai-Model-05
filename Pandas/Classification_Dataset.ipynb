```python
import torch
torch.__version__
torch.cuda.is_available()
torch.version.cuda
import numpy as np
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
```

```python
devide=torch.device("cpu")
if torch.cuda.is_available():
    device=torch.device("cuda")
elif torch.backends.mps.is_available():
    device=torch.device("mps")
    print("using device:",device)
```

```python
df=pd.read_csv("dm_office_sales.csv")
print(df.head())
print(df.shape)
```

```python
X=df.iloc[:, :-1]
Y=df.iloc[:, -1]

X=df.drop("sales",axis=1)
Y=df["sales"]
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

X=pd.get_dummies(X)

print(X)

X_train_df, X_test_df, Y_train, Y_test=train_test_split(X,Y,test_size=0.2,random_state=42)

# Initialize StandardScaler
scaler = StandardScaler()

# Fit on training data and transform both training and test data
X_train = scaler.fit_transform(X_train_df)
X_test = scaler.transform(X_test_df)

# X_train and X_test are now NumPy arrays, Y_train and Y_test are Series
print(X_test.shape, X_train.shape)
```

```python
X_train=torch.tensor(X_train.astype(np.float32),dtype=torch.float32)
X_test=torch.tensor(X_test.astype(np.float32),dtype=torch.float32)


Y_train=torch.tensor(Y_train.values.astype(np.float32),dtype=torch.float32).view(-1,1)
Y_test=torch.tensor(Y_test.values.astype(np.float32),dtype=torch.float32).view(-1,1)
print(X_train.shape)
```

```python
# defining model creating the model blueprint
from torch import nn
class linearRegression(nn.Module):
    def __init__(self):
        super().__init__()
        self.linear=nn.Linear(in_features=13,out_features=1)

    def forward(self,X): # tells when input comes pass it to the Linear function to tyrain Model
        return self.linear(X)

# create model
model=linearRegression()
loss_fn=torch.nn.MSELoss()
Optimizer=torch.optim.SGD(model.parameters(),lr=0.001,momentum=0.9)

# check prediction before training p[rovide prediction before learning this will provide random prediction ]
model.eval()
with torch.no_grad():
  Y_pred=model(X_test)
  print(Y_pred)

```

```python
# training loop Main learning part
for i in range(0,1000):
    Optimizer.zero_grad()
    outputs=model(X_train)
    loss=loss_fn(outputs,Y_train)
    loss.backward()
    Optimizer.step()
    if i%10==0:
        print(f'Epoch[{i+1}/1000],loss:{loss.item()}')

        # put model in evaluation

        model.eval()
        with torch.no_grad():
            Y_pred=model(X_test)
            print(Y_pred)
```
