```python
import pandas as pd
import numpy as np
import torch
import torch.nn as nn

from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split
```

```python
# Missing Values

print(iris.isnull().sum())
```

```python
# Remove unnecessary columns

if 'Id' in iris.columns:
    iris = iris.drop(columns=['Id'])

iris
```

```python
# Encoding (Categorical  -> Numeeric)

iris['Species'] = iris['Species'].astype('category').cat.codes

```

```python
iris
```

```python
# Feature and target split

X = iris.drop('Species',axis=1)
Y = iris['Species'].values

X.shape , Y.shape
```

```python
X , Y
```

```python
# Normalization

scalar_iris = MinMaxScaler()
X_scaled = scalar_iris.fit_transform(X)

X_scaled
```

```python
# Train-Test split

X_train, X_test , y_train , y_test = train_test_split(X_scaled,Y,test_size=0.2,random_state=42)

X_train , X_test , y_train , y_test
```

```python
# Convert to tensor

X_train = torch.tensor(X_train,dtype=torch.float32)
X_test = torch.tensor(X_test,dtype=torch.float32)
y_train = torch.tensor(y_train,dtype=torch.long)
y_test = torch.tensor(y_test,dtype=torch.long)

```

```python
# Model Definition

class IrisModel(nn.Module):
    def __init__(self):
        super(IrisModel,self).__init__()
        self.linear = nn.Linear(4,3)   # 4 features -> 3 classes

    def forward(self,x):
        return self.linear(x)
    
model = IrisModel()

```

```python
# Loss & optimizer

criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(),lr=0.01)

optimizer  , criterion
```

```python
# Training 

epochs = 1000

for epochs in range(epochs):
    outputs = model(X_train)
    loss = criterion(outputs, y_train)


    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    if (epochs+1) % 20 == 0:
        print(f"Epoch {epochs+1}, Loss:{loss.item(): .4f}")
```

```python
# Evaluation

with torch.no_grad():
    outputs = model(X_test)
    predicted = torch.max(outputs, 1)[1]

    accuracy = (predicted == y_test).sum().item() / len(y_test)

print("\n------- Predictions --------------")
print(predicted.numpy())

print("\n------- Actual ------------------")
print(y_test.numpy())

print(f"\n Accuracy: {accuracy * 100:.2f}%")
```
