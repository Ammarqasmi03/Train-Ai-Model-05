### Exceptions

If things go wrong during the execution of the program(runtime). It generally happens when something unforeseen has happened.

- Exceptions are raised by python runtime
- You have to takle is on the fly

#### **Examples**

- Memory overflow
- Divide by 0 -> logical error
- Database error

### Why is it important to handle exceptions?

- Prevents Crashes: Unhandled exceptions can cause a program to crash. By handling exceptions, you can prevent crashes and allow the program to continue running or exit gracefully.
- Provides User Feedback: Handling exceptions allows you to provide meaningful feedback to users when something goes wrong, rather than just showing a generic error message or stack trace.
- Allows for Recovery: In some cases, you may be able to recover from an exception and continue executing the program. For example, if a file is not found, you could prompt the user to provide a different file path.
- Improves Debugging: Properly handling exceptions can help you identify and fix bugs in your code more easily, as you can log the exception details or provide specific error messages that indicate what went wrong.

```python
# catching specific exceptions
try:
    with open('file.txt','r') as f:
        print(f.read())
    print(1/0)
except:
    print("An error occurred.")
```

```python
try:
    
    with open('file.txt','r') as f:
        print(f.read())
    k=10
    print(1/1)
    print(k)
    L = [1,2,3]
    L[100]
except FileNotFoundError:
    print("File not found. Please check the file name and try again.")
except NameError:
    print("variable not defined.")
except ZeroDivisionError:
    print("can't divide by zero.")
except Exception as e:
    print("An error occurred:", e)
```

```python
#else
try:
   f = open('simple.txt','r')
except FileNotFoundError:
    print("File not found. Please check the file name and try again.")
except Exception as e:
    print(e)
else:
    print(f.read())
    f.close()
```

```python
# Finally : 
# why to use finally block?
# The finally block is used to execute code that must be run regardless of whether an exception was raised or not. It is often used for cleanup actions, such as closing files or releasing resources.
try:
    with open('file.txt','r') as f:
        print(f.read())
    print(1/0)
except ZeroDivisionError:
    print("can't divide by zero.")
finally:
    print("This will always be executed.")
```

```python
# raise Exception
# In Python programming ,exceptions are raised when errors occur at runtime.
# We can also manually raise exceptions using the raise keyword. 

# raise -> In java throw
class bank:
    def __init__(self, balance):
        self.balance = balance
    
    def withdraw(self, amount):
        if amount < 0:
            raise Exception("Amount must be positive")
        if amount > self.balance:
            raise Exception("Insufficient funds")
        self.balance -= amount
        return self.balance
    
obj = bank(1000)
try:
    obj.withdraw(5000)
except Exception as e:
    print(e)
else:
    print("Withdrawal successful.")
    print("Remaining balance:", obj.balance)
```

#### Creating custom exceptions

- why we used this custom exception?
- Custom exceptions are useful for application based error and also full control

```python

class NegativeAmountError(Exception):
    def __init__(self, message):
        self.message = message
        super().__init__(self.message)
        


class bank:
    def __init__(self, balance):
        self.balance = balance
    
    def withdraw(self, amount):
        if amount < 0:
            raise NegativeAmountError("Amount must be positive you entered a -ve value try again!")
        if amount > self.balance:
            raise Exception("Insufficient funds")
        self.balance -= amount
        return self.balance
    
obj = bank(1000)
try:
    obj.withdraw(-5000)
except Exception as e:
    print(e)
else:
    print("Withdrawal successful.")
    print("Remaining balance:", obj.balance)
```

```python
# Simple Example:

class SecurityError(Exception):
    def __init__(self, message):
        print("Security Alert:", message)

    def logout(self):
        print("Logging out due to security concerns.")

class Google:
    
    def __init__(self,name,email,password,device):
        self.name = name
        self.email = email
        self.password = password
        self.device = device

    def login(self,email,password,device):
        if device !=self.device:
            raise SecurityError("Login attempt from unrecognized device.")
        if email == self.email and password == self.password:
            print("Login successful.")
        else:
            print("Invalid email or password.")

user = Google("nitish","nitish@gmail.com","nitish123","iPhone12")


try:
    user.login("nitish@gmail.com", "nitish123", "iPhone13")
except SecurityError as se:
    se.logout()
else:
    print(f"Welcome {user.name} to your account!")
finally:
    print("Thank you for using our service.")
```
