### Searching 

```python
# Linear Search

def linear_Search(arr,item):

    for i in range(len(arr)):

        if arr[i] == item:
            return i

    return -1
```

```python
arr = [2,4,16,5,67,1]

linear_Search(arr,67)
```

```python
# Binary Search

def binary_Search(arr,low,high,item):

    if low <= high:
        # mid 
        mid = (low+high)//2

        if arr[mid] == item:
            return mid
        elif arr[mid] > item:
            return binary_Search(arr,low,mid-1,item)
        else:
            return binary_Search(arr,mid+1,high,item)
    else:
        return -1
    
```

```python
arr = [12,34,56,78,90,99,100]

binary_Search(arr,0,len(arr)-1,120)
```

### Sorting 

```python
def is_sorted(arr):

    sorted = True

    for i in range(len(arr)-1):
        if arr[i]> arr[i+1]:
            sorted = False
            break

    return sorted
```

```python
arr = [1,2,3,4,8,6]

is_sorted(arr)
```

```python
import random

arr = [12,2,3,34,45,10]

random.shuffle(arr)
print(arr)
```

#### 2. Bubble Sort

- Worst Complexity : n^2
- Avg Complxty : n^2
- best Complxty : n
- swaping : n^2 for worst case
- comparaion : n^2 for worst case
- Space Complxty : 1
- Method : Exchanging
- Stable : Yes
- Class : Comparion Sort
- Adaptive : Yes

```python
def bubble_Sort(arr):

    for i in range(len(arr)-1):
        flag = 0
        for j in range(len(arr)-i-1):
            if arr[j] > arr[j+1]:
                arr[j],arr[j+1] = arr[j+1],arr[j]
                flag = 1
        if flag == 0:
            break
    return arr





```

```python
arr = [12,2,3,34,45,10]

bubble_Sort(arr)
```

#### 2. Selection Sort

- Worst Complexity : n^2
- Avg Complxty : n^2
- best Complxty : n^2 
- Swaping : n
- Comparion : n^2
- Space Complxty : 1
- Method : selection
- Stable : No
- Class : Comparion Sort
- Adaptive : No(Does not finish early if an array happens to already be sorted)
- **Benefits:**
    - less time taken than Bubble Sort
    - every iteration give the sub sorted array

```python
def selection_Sort(arr):

    for i in range(len(arr)):
        min_index = i
        for j in range(i+1,len(arr)):
            if arr[min_index] > arr[j]:
                min_index = j
        if min_index != i:
            arr[i],arr[min_index] = arr[min_index],arr[i]

    return arr
```

```python
arr = [8,9,3,1,7]

selection_Sort(arr)
```

#### Merge Sort

- Worst Complexity : nlog(n)
- Avg Complxty : nlog(n)
- best Complxty : nlog(n)
- Space Complxty : n
- Method : merging
- Stable : Yes
- Class : Merged Sort
- Adaptive : No(Does not finish early if an array happens to already be sorted)

```python
def merge_sorted(arr1,arr2,arr):
    i = j = k =  0
    
    while i < len(arr1) and j < len(arr2):
        if arr1[i] < arr2[j]:
            arr[k] = arr1[i]
            i += 1
        else:
            arr[k] = arr2[j]
            j += 1
        k+=1


    while i < len(arr1):
        arr[k] = arr1[i]
        i += 1
        k+=1

    while j < len(arr2):
        arr[k] = arr2[j]
        j += 1
        k+=1

    
```

```python
def merge_sort(arr):

    if len(arr) == 1:
        return arr
    
    mid = len(arr) //2

    left =  arr[:mid]
    right = arr[mid:]

    merge_sort(left)
    merge_sort(right)

    merge_sorted(left,right,arr)
```

```python
arr = [2,4,7,5,3,1,9,0,8]
merge_sort(arr)

print(arr)
```
