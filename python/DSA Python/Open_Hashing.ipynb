```python
class Dictionary:

    def __init__(self,size):
        self.size = size
        self.keys = [None] * size
        self.data = [None] * size

    def __str__(self):
        
        for i in range(len(self.keys)):
            if self.keys[i] != None:
                print(self.keys[i],':',self.data[i],end=',')
        return ""
    
    def Size(self):
        return self.size
    
    def put(self,key,value):
        hash_value = self.hash_Function(key)
        if self.keys[hash_value] == None:
            self.keys[hash_value] = key
            self.data[hash_value] = value
        else:
            if self.keys[hash_value] == key:
                self.data[hash_value] = value
            else:
                new_hash_value = self.rehash(hash_value)
                while self.keys[new_hash_value] != None  and self.keys[new_hash_value] != key:
                    new_hash_value = self.rehash(new_hash_value)

                if self.keys[new_hash_value] == None:
                    self.keys[new_hash_value] = key
                    self.data[new_hash_value] = value 
                else:
                    self.data[new_hash_value] = value

    def get(self,key):
        start_pos = self.hash_Function(key)
        curr_pos = start_pos

        while self.keys[curr_pos] != None:

            if self.keys[curr_pos] == key:
                return self.data[curr_pos]
            
            curr_pos = self.rehash(curr_pos)
            
            if curr_pos == start_pos:
                return 'Not Found'
            
        return 'Nahi Wala None'

    def __getitem__(self, key):
        return self.get(key)

    def __setitem__(self, key, value):
        self.put(key,value)
                         

    def rehash(self,old_hash_value):
        return (old_hash_value + 1)  % self.size

    def hash_Function(self,key):
        return  abs(hash(key)) % self.size   
        
```
