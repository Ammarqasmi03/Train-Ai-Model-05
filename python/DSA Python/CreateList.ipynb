```python
import ctypes
```

```python
class myList:
    # Create List
    def __init__(self):
        self.n = 1
        self.size = 0
        # Create a Ctypes array n = self.n
        self.A = self.__make_array(self.n)

    
    
    # length method
    def __len__(self):
        return self.size

    # append method    
    def append(self,item):
        if self.n == self.size:
            # resize
            self.__resize(self.n*2)
        self.A[self.size] = item
        self.size = self.size + 1

    

    # print method
    def __str__(self):
        # [1,2,3]
        result = ''
        for i in range(self.size):
            result = result + str(self.A[i]) + ','

        return '[' + result[:-1] + ']'
    
    # Indexing Method

    def __getitem__(self, key):
        if 0 <= key < self.size:
            return self.A[key]
        elif  -self.size <= key < 0:
            return self.A[key+self.size]
        else:
            return 'IndexError - Index out of range' 


    # slicing 
    def __getitem__(self, key):
        if isinstance(key,slice):
            result = []
            for i in range(*key.indices(self.size)):
                result.append(self.A[i])
            return result    

    # Pop method : delete the last item of the list

    def pop(self):
        if self.size == 0:
            return 'Empty List'
        print(self.A[self.size -1])
        self.size = self.size - 1


    # clear method : delete all the element of the list

    def clear(self):
        self.size = 0
        self.n = 1

            
    # find method : get the index of the items

    def find(self,item):
        for i in range(self.size):
            if item == self.A[i]:
                return i
            
        return f'{item} is not in List'

    # insert method : to insert the item at the given index pos        
    
    def insert(self,index,item):
        if self.size == self.n:
            self.__resize(self.n*2)
        for i in range(self.size,index,-1):
            self.A[i] = self.A[i-1]
        self.A[index] = item
        self.size = self.size + 1
    
    # delete Method : to delete the item from the given index 

    def delete(self,index):
        if 0<= index < self.size:
            
            for i in range(index+1,self.size):
               self.A[i-1] = self.A[i]
            self.size = self.size - 1
        
    def remove(self,item):
        pos = self.find(item)
        if type(pos) == int:
            self.delete(pos)
        else: 
            return pos


    # max method : find the max element

    def max(self):
        max_elem = self.A[0]

        for i in range(1, self.size):
            if self.A[i] > max_elem:
                max_elem = self.A[i]

        print(max_elem)


    # min method : find the minimum element

    def min(self):
        min_elem = self.A[0]

        for i in range(1,self.size):
            if self.A[i] < min_elem:
                min_elem = self.A[i]
        print(min_elem)

        
    # sort method : To sort the list element

    def sort(self):

       for i in range(self.size - 1):
           min_index = i

           for j in range(i + 1, self.size):
               if self.A[j] < self.A[min_index]:
                  min_index = j

           if min_index != i:
               self.A[i], self.A[min_index] = self.A[min_index], self.A[i]

    # Sum method : to sum all the element of the list       
    def sum(self):
        result = 0
        for i in range(self.size):
            result = result + self.A[i]

        print(result)
    
    # extend method : use to append the new list on the old list

    def extend(self,list):
        for i in range(len(list)):
            if self.size == self.n:
                self.__resize(self.size*2)
            self.A[self.size] = list[i]
            self.size += 1
            

     

    
   
        




    def __resize(self,new_capacity):
        # create a new array with new capacity 
        B = self.__make_array(new_capacity)
        self.n = new_capacity
        # copy the content of A to B
        for i in range(self.size):
            B[i] = self.A[i]

        # reassign A
        self.A = B

    def __make_array(self,capacity):
        # creates a C type array(static,referential) with size capacity

        return (capacity*ctypes.py_object)()
```
