```python
class Node:

    def __init__(self,value):
        self.data = value
        self.next = None
        
```

```python
class LinkedList:

    def __init__(self):
        self.head = None
        # No of Nodes
        self.size = 0

    def __len__(self):
        return self.size
    
    # insert data from head 
    def insert_head(self,value):
        # new Node
        new_Node = Node(value)

        # create connection
        new_Node.next = self.head

        # reassign head
        self.head = new_Node

        #increment size
        self.size += 1
    
    # insert data from tail
    def insert_tail(self,value):
        new_Node = Node(value)
        # LL is empty
        if self.head == None:
            self.head = new_Node
            self.size += 1
            return

        curr = self.head
        while curr.next != None:
              curr = curr.next
        # you are at the last Node
        curr.next = new_Node
        self.size += 1


    def insert_After(self,after,value):
      
        new_Node = Node(value)
        curr = self.head
        while curr != None:
             if curr.data == after:
                 break
             curr = curr.next
              
        if curr != None:
            new_Node.next = curr.next
            curr.next = new_Node
            self.size += 1
        else:
            return f'{after} is not in list'

    def clear(self):
        self.head = None
    
    def delete_head(self):
        if self.head == None:
            return 'Empty LL'
        self.head = self.head.next
        self.size -= 1

    def pop(self):
        
        # No element is left i.e list is empty
        if self.head == None:
            return 'Empty LL'
        
        curr = self.head
        
        # single element is left
        
        if curr.next == None:
            return self.delete_head()

        while curr.next.next != None:
            curr = curr.next

        curr.next = None
        self.size -= 1

    def remove(self,value):

        if self.head == None:
            return 'Empty LL'
         
        curr = self.head
        
        if curr.data == value:
            return self.delete_head()

        while curr.next != None:
            if curr.next.data == value:
                break
            curr = curr.next

        if curr.next != None:
            curr.next = curr.next.next
            self.size -= 1
        else:
            return f'{value} is not in list'
        
    def __delitem__(self, index):
        curr = self.head
        pos = 0
            
        while curr != None:
            if pos == index:
                value = self.__getitem__(index)
                print(value)
                return self.remove(value)
            curr = curr.next
            pos += 1
        print('IndexError')
        
    def search(self,value):
        curr = self.head
        index = 0
        
        while curr != None:
            if curr.data == value:
                return index
            index += 1
            curr = curr.next


        return 'Not Found'
            

    def __getitem__(self, index):
        
        curr = self.head
        pos = 0

        while curr != None:
            if pos == index:
                return curr.data 
            curr = curr.next
            pos += 1

        return 'IndexError'
       


    # To print the data in LinkedList
    def __str__(self):

        curr = self.head

        result = ''
        while curr != None:
            result = result + str(curr.data) + '->'
            curr = curr.next
 
        return result[:-2]

```
