```python
class Node:

    def __init__(self,value):
        self.data = value
        self.next = None
```

```python
class Stack:

    def __init__(self):
        self.top = None
        self.size = 0

    def isEmpty(self):
        return self.top == None
    
    def push(self,value):

        new_Node = Node(value)

        new_Node.next = self.top

        self.top = new_Node

        self.size += 1

    def peep(self):
        if self.top == None:
            return 'Empty Stack' 
        else:
            return self.top.data
        
    def pop(self):
        if self.top == None:
            return 'Empty Stack'
        else:
            data = self.top.data
            self.top = self.top.next
            self.size -= 1
            return data
        
    def Size(self):
        
        return self.size
        
    def traverse(self):

        temp = self.top

        while temp != None:
            print(temp.data)
            temp = temp.next

            
```

```python
# Reverse the string

def reverse_String(text):
    s = Stack()
    
    for i in text:
        s.push(i)

    res = ''

    while not s.isEmpty():
        res = res + s.pop()

    return res
```

```python
# text Editor

def text_Editor(text,pattern):

    s1 = Stack()
    s2 = Stack()
    for i in text:
        s1.push(i)


    for i in pattern:
        if i == 'u':   # undo or pop()
            un = s1.pop()
            s2.push(un)
        else:          # redo or push()
            re = s2.pop()
            s1.push(re)

    res = ''
    while not s1.isEmpty():
         res = s1.pop() + res

    return res
    
```

```python
# Bracket Matched

def check_parenthesis(equation):

    s = Stack()

    pairs = {
        ')' : '(',
        '}' : '{',
        ']' : '['

    }
    
    for i in equation:
        if i in '{[(':
           s.push(i)
           
        elif i in ')]}':
            if s.Size() == 0:
                return False
            if s.peep() == pairs[i]:
                s.pop()
            else:
                return False
        

    
    return s.Size() == 0
    
   
    

            




```

```python
equation = '[(a+b)]'

check_parenthesis(equation)
```

#### Queue

```python
class Queue:

    def __init__(self):
        self.front = None
        self.rear = None

    def isEmpty(self):
        return self.rear == None
    
    def Size(self):

        temp = self.front
        counter = 0

        while temp != None:
            temp = temp.next
            counter += 1

        return counter
    

    def traverse(self):

        temp = self.front

        while temp != None:
            print(temp.data,end=' ')
            temp = temp.next

    def enqueue(self,value):

        new_Node = Node(value)
        
        if self.rear == None:
            self.front = new_Node
            self.rear = self.front
        else:
            self.rear.next = new_Node
            self.rear = new_Node

    def dequeue(self):

        if self.front == None:
            return 'Empty'
        else:
            self.front = self.front.next

    def front_item(self):
        if self.front == None:
            return 'empty'
        else:
            return self.front.data
        
    def rear_item(self):
        if self.front == None:
            return 'empty'
        else:
            return self.rear.data
```
