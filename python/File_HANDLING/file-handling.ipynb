### Writing to a file

```python
# case1 - if the file is not present
f = open('simple.txt','w')
f.write('This is a simple file')
f.close()
# since file is closed, we cannot write to it
f.write('This is a simple file')
```

```python
 # write multiline strings
f = open('simple1.txt','w')
f.write('This is a simple file\n')
f.write('This is a second line\n')
f.write('This is a third line\n')
f.close()
```

```python
# case 3 - if the file is already present
f = open('simple.txt','w')
f.write('This is a new file')
f.close()
```

```python
# how exactly open() works?
# open() takes two arguments: filename and mode
# mode can be 'r' (read), 'w' (write), 'a' (append)
# if file doesn't exist and mode is 'w', it creates a new file
# if file exists and mode is 'w', it overwrites the existing file
# if file doesn't exist and mode is 'r', it raises an error
# if file exists and mode is 'r', it reads the existing file

# Problem with w mode
# introducing append 'a' mode
f = open('simple1.txt','a')  
f.write('This is a fourth line\n')
f.close()
```

```python
# Writes lines
L = ['Hello\n','Hi\n','How are you?\n','This is a new line\n']
f = open('simple.txt','w')
f.writelines(L)    # writelines() takes a list of strings and writes them to the file
f.close()
```

```python
 # Reading from files
# using read() method
f = open('simple.txt','r')
data = f.read()   # read() reads the entire file and returns it as a string
print(data)
f.close()  
```

```python
# readline() method reads a single line from the file and returns it as a string
f = open('simple.txt','r')
line1 = f.readline()  # reads the first line
line2 = f.readline()  # reads the second line
print(line1 ,end='')
print(line2,end='')
f.close()
```

```python
# Reading upto n chars
f = open('simple.txt','r')
data = f.read(10)   # read() reads the first 10 characters of the file and returns it as a string
print(data)
f.close()
```

```python
# Raeding entrie using readline() method

f = open('simple.txt','r')

while True:
    line = f.readline()
    if not line:  # if line is empty, we have reached the end of the file
        break
    print(line,end='')

f.close()
```

### Using Context Manager(With)

```python
# With statement

# with statement is used to automatically close the file after the block of code is executed

# try f.read() now without closing the file
with open('simple.txt','r') as f:
    data = f.read()
    print(data)
```

```python
# moving within a file -> 10 chars then 10 chars
with open('simple.txt','r') as f:
    print(f.read(9),end='')  # read first 9 chars
    print(f.read(12),end='')  # read next 12 chars
    print(f.read(20))  # read next 10 chars
```

```python
# benefits -> to load a big file in memory

big_L = ['Hello World ']*1000   # creates a list of 1000 strings


with open('big.txt','w') as f:
    f.writelines(big_L)  # writes the list of strings to the file

    
```

```python
with open('big.txt','r') as f:

    chunk_size = 10  # read 100 characters at a time

    while len(f.read(chunk_size)) > 0:
        print(f.read(chunk_size),end='')
        f.read(chunk_size)
```

```python
# seek and tell methods

# seek : moves the file pointer to a specific position in the file
# tell : returns the current position of the file pointer

# seek() takes two arguments: offset and whence
# offset is the number of bytes to move the file pointer
# whence is the reference point from where the offset is calculated
# whence can be 0 (default, beginning of the file), 1 (current
# position of the file pointer), 2 (end of the file)

with open('simple.txt','r') as f:
    print(f.read(10))  # read first 10 chars
    print(f.tell())  # returns the current position of the file pointer
    f.seek(0)  # move the file pointer to the beginning of the file
    print(f.read(10))  # read first 10 chars again
    f.seek(5)  # move the file pointer to the 5th byte from the beginning of the file

```

```python
# seek during writing to a file
with open('simple.txt','w') as f:
    f.write('Hello World')
    f.seek(0)  # move the file pointer to the beginning of the file
    f.write('Hi ')  # overwrite the first two characters of the file
```

#### Serialization and Deserialization


```python
 # serialization using json module
#list
import json

L = [1,2,3,4,5]

with open('list.json','w') as f:
    json.dump(L,f)  # dump() method takes a Python object and a file object and writes the JSON representation of the object to the file
```

```python
# dictionary

D = {'name':'John','age':30,'city':'New York'}

with open('dict.json','w') as f:
    json.dump(D,f,indent=4)  # dump() method takes a Python object and a file object and writes the JSON representation of the object to the file
```

```python
# deserialization using json module

import json

with open('dict.json','r') as f:
    D = json.load(f) 
    # load() method takes a file object and returns the JSON representation of the file as a Python object
    print(D)
    print(type(D))
```

```python
# serialize and deserialize tuple
import json

T = (1,2,3,4,5)

with open('demo.json','w') as f:
    json.dump(T,f)  # dump() method takes a Python object and a file object and writes the JSON representation of the object to the file
```

### Serialinzing and deserializing custom object

```python
class Person:
    def __init__(self,fname,lname,age,gender):
        self.fname = fname
        self.lname = lname
        self.age = age
        self.gender = gender


person = Person('John','Doe',30,'Male')
```

```python
# As a string 

import json

def show_object(person):
    if isinstance(person, Person):
        return "fullNAME -> {} {} age -> {}  gender -> {}".format(person.fname,person.lname,person.age,person.gender)
    

with open('demo.json','w') as f:
    json.dump(person,f,default=show_object)  # dump() method takes a Python object and a file object and writes the JSON representation of the object to the file
```

```python
# As a dictionary

import json

def show_object(person):
    if isinstance(person, Person):
        return {'fname': person.fname, 'lname': person.lname, 'age': person.age, 'gender': person.gender}
    
with open('demo.json','w') as f:
    json.dump(person,f,default=show_object,indent=4)  # dump() method takes a Python object and a file object and writes the JSON representation of the object to the file
```

```python
# deserialization of custom objects

import json

with open('demo.json','r') as f:
    print(json.load(f))  # load() method takes a file object and returns the JSON representation of the file as a Python object
```

### Pickling

##### Pickling is the process whereby a python object hierarchy is converted into a byte stream



```python
class Person:

    def __init__(self,name,age):
        self.name = name
        self.age = age

    def display_info(self):
        print('Hi my name is {} and I am {} years old'.format(self.name,self.age))


p = Person('John',30)
```

```python
# pickle dump
import pickle

with open('person.pkl','wb') as f:
    pickle.dump(p,f)  # dump() method takes a Python object and a file object and writes the binary representation of the object to the file
```

```python
# pickle load
import pickle
with open('person.pkl','rb') as f:
    p1 = pickle.load(f)  # load() method takes a file object and returns the binary representation of the file as a Python object
    print(p1)
    print(type(p1))
    p1.display_info()
```

### Pickle vs Json

##### -> Pickle lets the user to store data in binary format. JSON lets the user store data in a human readable text format
