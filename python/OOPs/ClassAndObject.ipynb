```python
class Person:
    def __init__(self):
        self.name = 'Ammar'
        self.gender = 'male'
        
# Object without a reference variable
Person()
```

```python
# Multiple reference variables pointing to the same object

p = Person()
q = p

print(id(q))
print(id(p))

print(p.name)
print(q.name)
```

## Pass by reference

```python
class Person:
    def __init__(self,name,gender):
        self.name = name
        self.gender = gender

# Outside the class -> function
def greet(person):
    print(f"Hello, {person.name}! You are a {person.gender}.")


p = Person("Ammar", "male")
greet(p)
```

### Object is mutable


```python
class Person:

  def __init__(self,name,gender):
    self.name = name
    self.gender = gender

# outside the class -> function
def greet(person):
  person.name = 'ankit'
  return person

p = Person('nitish','male')
print(id(p))
p1 = greet(p)
print(id(p1))
```
