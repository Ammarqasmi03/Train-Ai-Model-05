## Aggreagation

```python
class Customer:

    def __init__(self,name,gender,address):
        self.name = name
        self.gender = gender
        self.address = address

    def print_Address(self):
        print("City:",self.address.get_City())
        print("Pin:",self.address.pin)
        print("State:",self.address.state)

    def edit_profile(self,new_name,new_city,new_pin,new_state):
        self.name = new_name
        self.address.edit_Address(new_city,new_pin,new_state)
        


class Address:
    
    def __init__(self,city,pin,state):
        self.pin = pin
        self.__city = city
        self.state = state

    def get_City(self):
        return self.__city

    def edit_Address(self,new_city,new_pin,new_state):
        self.__city = new_city
        self.pin = new_pin
        self.state = new_state

a1 = Address("Bangalore",560037,"Karnataka")
c1 = Customer("Ravi","Male",a1)

c1.print_Address()

c1.edit_profile("Ravi Kumar","Mysore",570001,"Karnataka")
print("\nAfter editing profile:\n")
c1.print_Address()
```

## Inheritance

### Example

```python
# Parent 

class User:

    def __init__(self):
        self.name = "Ravi"

    def login(self):
        print("User logged in")

# Child class inheriting from User

class Student(User):

    def __init__(self):
        super().__init__()  # Call the constructor of the parent class
        self.student_id = "12345"

    def display_info(self):
        print("Name:", self.name)
        print("Student ID:", self.student_id)


u = User()
s = Student()

s.login()  # Inherited method from User
s.display_info()  # Method from Student
```

```python
# Child can't access private members of parent class

class Phone:

    def __init__(self, brand, price, camera):
        print("Inside phone constructor")
        self.brand = brand
        self.camera = camera
        self.__price = price  # Private member

    def __Show(self):
        print(self.__price)  # Accessing private member within the class

class SmartPhone(Phone):

    def check(self):
        # print(self.__price)  # This will raise an error as __price is private to Phone class
        pass

s1 = SmartPhone("Apple", 100000, "12MP")
print(s1.brand)
s1.check()
s1.Show()
```

```python
# Method Overriding

class Phone:

    def __init__(self, brand, price, camera):
        print("Inside phone constructor")
        self.brand = brand
        self.__price = price
        self.camera = camera

    def buy(self):
        print("Buying a phone")

class SmartPhone(Phone):

    def buy(self):
        print("Buying a smartphone")

s1 = SmartPhone("Apple", "iPhone 12", "12MP")
s1.buy()  # This will call the buy method of SmartPhone, not Phone
```

#### Super Keyword

```python
class Phone:
    def __init__(self, brand, price, camera):
        print("Inside phone constructor")
        self.brand = brand
        self.__price = price
        self.camera = camera

    def buy(self):
        print("Buying a phone")

class SmartPhone(Phone):
 
    def __init__(self, brand, price, camera, os, ram):
        print("Inside smartphone constructor")
        super().__init__(brand, price, camera)  # Call the constructor of the parent class
        self.os = os
        self.ram = ram
        print("Smartphone details initialized")
        

    def buy(self):
        print("Buying a smartphone")
        super().buy()  # Call the buy method of the parent class
        

s1 = SmartPhone("Apple", 100000, "12MP", "iOS", "4GB")
s1.buy()  # This will call the buy method of SmartPhone, which also calls the buy method of Phone

# using the super() outside the class is not valid
# EX: super().buy()  # This will raise an error as super() cannot be used outside of a class method

# Super cannot access the variables of the parent class
#Ex: super().brand  # This will raise an error as super() cannot access instance variables of the parent class directly
```

## Inheritance in summary

#### A class can inherit from another class.
#### Inheritance improves code reuse
#### Constructor, attributes, methods get inherited to the child class
#### The parent has no access to the child class
#### Private properties of parent are not accessible directly in child class
#### Child class can override the attributes or methods. This is called method overriding
#### super() is an inbuilt function which is used to invoke the parent class methods and constructor

## Types of Inheritance:

#### -> Single
#### -> Multilevel
#### -> Hierarchial
#### -> Multiple
#### -> Hybrid

```python
# single inheritance
class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")

class SmartPhone(Phone):
    pass

SmartPhone(1000,"Apple","13px").buy()
```

```python
# multilevel
class Product:
    def review(self):
        print ("Product customer review")

class Phone(Product):
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")

class SmartPhone(Phone):
    pass

s=SmartPhone(20000, "Apple", 12)

s.buy()
s.review()
```

```python
# Hierarchical
class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")

class SmartPhone(Phone):
    pass

class FeaturePhone(Phone):
    pass

SmartPhone(1000,"Apple","13px").buy()
FeaturePhone(10,"Lava","1px").buy()
```

```python
# Multiple
class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")

class Product:
    def review(self):
        print ("Customer review")

    def buy(self):
        print("Buying a product")

class SmartPhone(Product, Phone):
    pass

s=SmartPhone(20000, "Apple", 12)

s.buy()
s.review()
```

```python
# the diamond problem
# https://stackoverflow.com/questions/56361048/what-is-the-diamond-problem-in-python-and-why-its-not-appear-in-python2
class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")

class Product:
    def buy(self):
        print ("Product buy method")

# Method resolution order
class SmartPhone(Phone,Product):
    pass

s=SmartPhone(20000, "Apple", 12)

s.buy()
```

# PolyMorphism

##### -> Method Overriding
##### -> Method Overloading
##### -> Operator OverLoading

```python
# Method Overloading

class shape:
        
    def area(self, a, b=0):
        if b == 0:
            print("Area of square:", a * a)
        else:
            print("Area of rectangle:", a * b)
   

s = shape()
s.area(5)  # This will call the area method for square, not circle or rectangle
s.area(5, 10)  # This will call the area method for square, not circle or rectangle
```
