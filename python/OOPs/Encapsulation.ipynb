### Encapsulation


```python
class Atm:

    def __init__(self):
        self.pin = ' '
        self.__balance = 0

    def get_balance(self):
        return self.__balance

    def set_balance(self,new_value):
        if type(new_value) == int:
           self.__balance = new_value
        else:
           print('beta bahot maarenge')
```

```python
# List of objects

class Person:
    
    def __init__(self,name,age):
        self.name = name
        self.age = age

p1 = Person('Alice', 30)
p2 = Person('Bob', 25)
p3 = Person('Charlie', 35)

people = [p1, p2, p3]

for person in people:
    print(f'Name: {person.name}, Age: {person.age}')



# Dict of objects

class Student:
    
    def __init__(self,name,grade):
        self.name = name
        self.grade = grade

s1 = Student('David', 'A')
s2 = Student('Eve', 'B')
s3 = Student('Frank', 'C')

students = {'s1': s1, 's2': s2, 's3': s3}

for key, student in students.items():
    print(f'Key: {key}, Name: {student.name}, Grade: {student.grade}')
```

### Static Variables   &&  Static Method

```python
class Atm:

    __counter = 1


    def __init__(self):
        self.pin = ' '
        self.__balance = 0
        self.cid = Atm.__counter
        Atm.__counter = Atm.__counter + 1


    def get_balance(self):
        return self.__balance

    def set_balance(self,new_value):
        if type(new_value) == int:
           self.__balance = new_value
        else:
           print('beta bahot maarenge')

    @staticmethod
    def get_counter():
        return Atm.__counter
```
