```python
import requests
import pandas as pd
import time

session = requests.Session()

headers = {
    "accept": "application/json",
    "Authorization": "Bearer eyJhbGciOiJIUzI1NiJ9.eyJhdWQiOiI1YWU2YjVhMDlkNzU2NzgwZjZjMWZlN2RjZmE4MmJiMCIsIm5iZiI6MTc4MTUzNDEyNy40Niwic3ViIjoiNmEzMDBkYWY4NmFlNzJkMWNmNzNhN2U2Iiwic2NvcGVzIjpbImFwaV9yZWFkIl0sInZlcnNpb24iOjF9.X-G5ALYuOoFKQJa_NUtr82TCynnQHZMdcoZR95GrK1Y"
}

all_movies = []

for page in range(1, 548):

    url = f"https://api.themoviedb.org/3/movie/top_rated?page={page}"

    success = False

    for attempt in range(3):

        try:
            response = session.get(
                url,
                headers=headers,
                timeout=30
            )

            data = response.json()

            all_movies.extend(data['results'])

            print(f"Downloaded page {page}")

            success = True
            break

        except Exception as e:

            print(
                f"Page {page}, "
                f"Attempt {attempt+1} failed"
            )

            time.sleep(2)

    if not success:
        print(f"Skipping page {page}")

    time.sleep(0.2)

df = pd.DataFrame(all_movies)

print("Total Movies:", len(df))
```

```python
temp_df = df[
    ['id','title','overview',
     'release_date','popularity',
     'vote_average','vote_count']
]
```

```python
temp_df.to_csv('top_rated_movies.csv', index=False)
```
