```python
import pandas as pd
```

### 1. to_csv

```python
df = pd.read_csv(r'D:\mca_aiml_lab\mca_env\Datasets\deliveries.csv')
df.head()
```

```python
temp_df = df.groupby('batsman')['batsman_runs'].sum().reset_index()
temp_df
```

```python
temp_df.to_csv('Batsman_runs.csv',index=False)
```

```python
temp_df2 = df.pivot_table(index='batsman',columns='bowling_team',values='batsman_runs',aggfunc='sum')
```

###  to_html

```python
ipl_sixes = df.query('batsman_runs == 6').pivot_table(index='over',columns='ball',values='batsman_runs',aggfunc='count')
```

```python
ipl_sixes
```

```python
ipl_sixes.to_html('sixes_heatmap.html')
```

### to_json

```python
team_player_run = df.groupby(['batting_team','batsman'])['batsman_runs'].sum().unstack().unstack().unstack()
```

```python
team_player_run.to_json('ipl.json')
```
